```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0, 2 * np.pi)
y = np.sin(x)
```

```python
plt.plot(x, y);
```

```python
from matplotlib import style
style.use('ggplot')
```

```python
x = np.arange(20)
y = x % 5
y2 = x // 5
plt.plot(x, y)
plt.plot(x, y2)
plt.title("Resto da divisão por 5")
plt.xlabel("Dividendo")
plt.ylabel("Resto");
```

```python
paises = ['Eua', 'Ind', 'Bra', 'Fra', 'Rus', 'Esp', 'Arg', 'Run', 'Col', 'Mex']
casos = [10.02, 8.55, 5.66, 1.84, 1.76, 1.33, 1.24, 1.2, 1.14, 0.97]
ativos = [5.82, 0.51, 0.35, 1.66, 0.41, 1.14, 0.15, 1.14, 0.07, 0.05]

plt.bar(paises, casos, color='b', label="Confirmados");
plt.bar(paises, ativos, color='y', label="Ativos", alpha=0.5);
plt.legend();
```
