```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
from mpl_toolkits.mplot3d import Axes3D
from matplotlib.animation import FuncAnimation

def plot_area_under_curve(func, x_range):
    """Plota a área sob a curva."""
    x_vals = np.linspace(x_range[0], x_range[1], 100)
    y_vals = func(x_vals)

    plt.figure(figsize=(12, 6))
    plt.subplot(1, 2, 1)
    plt.plot(x_vals, y_vals, label='f(x)', color='blue')
    plt.fill_between(x_vals, y_vals, color='lightblue', alpha=0.5)
    
    plt.title('Área sob a Curva')
    plt.xlabel('x')
    plt.ylabel('f(x)')
    plt.axhline(0, color='black', lw=0.5)
    plt.axvline(0, color='black', lw=0.5)
    plt.legend()

def plot_solid_of_revolution(func, x_range):
    """Plota o sólido de revolução e anima a rotação."""
    # Criar um intervalo de x
    x_vals = np.linspace(x_range[0], x_range[1], 100)
    y_vals = func(x_vals)

    # Criar uma grade para o sólido
    theta = np.linspace(0, 2 * np.pi, 100)
    X, T = np.meshgrid(x_vals, theta)
    
    # Coordenadas Y e Z do sólido
    Y = func(X) * np.cos(T)  # Coordenada Y
    Z = func(X) * np.sin(T)  # Coordenada Z

    fig = plt.figure(figsize=(12, 6))
    ax = fig.add_subplot(122, projection='3d')
    
    # Plotar o sólido
    ax.plot_surface(X, Y, Z, alpha=0.6)

    # Configurações do gráfico 3D
    ax.set_title('Sólido de Revolução')
    ax.set_xlabel('X')
    ax.set_ylabel('Y')
    ax.set_zlabel('Z')

def animate_solid(func_np, x_range):
    """Anima o sólido de revolução girando em torno do eixo X."""
    
    fig = plt.figure(figsize=(8, 8))
    
    ax = fig.add_subplot(111, projection='3d')
    
    # Criar um intervalo de x
    x_vals = np.linspace(x_range[0], x_range[1], 100)
    
    theta = np.linspace(0, 2 * np.pi, 100)  # Ângulo para a rotação
    
    def update(frame):
        ax.clear()
        Y = func_np(x_vals) * np.cos(theta + frame / 10.0)  # Rotação em torno do eixo X
        Z = func_np(x_vals) * np.sin(theta + frame / 10.0)

        ax.plot_surface(x_vals[:, None], Y.T, Z.T, alpha=0.5)

        ax.set_title('Sólido de Revolução (Animação)')
        ax.set_xlabel('X')
        ax.set_ylabel('Y')
        ax.set_zlabel('Z')
        ax.set_xlim([x_range[0], x_range[1]])
        ax.set_ylim([-2, 2])
        ax.set_zlim([-2, 2])

    ani = FuncAnimation(fig, update, frames=np.arange(0, 360), interval=50)
    
    plt.show()

def main():
    print("Calculadora de Sólidos de Revolução")
    
    # Definir a variável simbólica
    x = sp.symbols('x')

    # Inserir manualmente a função que deseja integrar
    func_input = input("Digite a função f(x) para integrar (ex: sqrt(sin(x)) + cos(x)**2): ")
    
    try:
        func_sympy = sp.sympify(func_input)
        func_np = sp.lambdify(x, func_sympy, 'numpy')  # Converter para função NumPy
    except sp.SympifyError as e:
        print(f"Erro na entrada da função: {e}")
        return

    # Definir o intervalo de integração
    lower_limit = float(input("Digite o limite inferior da integral: "))
    upper_limit = float(input("Digite o limite superior da integral: "))
    
    plot_area_under_curve(func_np, (lower_limit, upper_limit))
    
    plot_solid_of_revolution(func_np, (lower_limit, upper_limit))
    
    animate_solid(func_np, (lower_limit, upper_limit))

if __name__ == "__main__":
    main()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
from mpl_toolkits.mplot3d import Axes3D
from matplotlib.animation import FuncAnimation

def plot_area_under_curve(func, x_range):
    """Plota a área sob a curva."""
    x_vals = np.linspace(x_range[0], x_range[1], 100)
    y_vals = func(x_vals)

    plt.figure(figsize=(12, 6))
    plt.subplot(1, 2, 1)
    plt.plot(x_vals, y_vals, label='f(x)', color='blue')
    plt.fill_between(x_vals, y_vals, color='lightblue', alpha=0.5)
    
    plt.title('Área sob a Curva')
    plt.xlabel('x')
    plt.ylabel('f(x)')
    plt.axhline(0, color='black', lw=0.5)
    plt.axvline(0, color='black', lw=0.5)
    plt.legend()

def plot_solid_of_revolution(func, x_range):
    """Plota o sólido de revolução."""
    # Criar um intervalo de x
    x_vals = np.linspace(x_range[0], x_range[1], 100)
    y_vals = func(x_vals)

    # Criar uma grade para o sólido
    theta = np.linspace(0, 2 * np.pi, 100)
    X, T = np.meshgrid(x_vals, theta)
    
    # Coordenadas Y e Z do sólido
    Y = func(X) * np.cos(T)  # Coordenada Y
    Z = func(X) * np.sin(T)  # Coordenada Z

    fig = plt.figure(figsize=(12, 6))
    ax = fig.add_subplot(122, projection='3d')
    
    # Plotar o sólido
    ax.plot_surface(X, Y, Z, alpha=0.6)

    # Configurações do gráfico 3D
    ax.set_title('Sólido de Revolução')
    ax.set_xlabel('X')
    ax.set_ylabel('Y')
    ax.set_zlabel('Z')

def animate_solid(func_np, x_range):
    """Anima o sólido de revolução girando em torno do eixo X."""
    
    fig = plt.figure(figsize=(8, 8))
    
    ax = fig.add_subplot(111, projection='3d')
    
    # Criar um intervalo de x
    x_vals = np.linspace(x_range[0], x_range[1], 100)
    
    theta = np.linspace(0, 2 * np.pi, 100)  # Ângulo para a rotação
    
    def update(frame):
        ax.clear()
        Y = func_np(x_vals) * np.cos(theta + frame / 10.0)  # Rotação em torno do eixo X
        Z = func_np(x_vals) * np.sin(theta + frame / 10.0)

        ax.plot_surface(x_vals[:, None], Y.T, Z.T, alpha=0.5)

        ax.set_title('Sólido de Revolução (Animação)')
        ax.set_xlabel('X')
        ax.set_ylabel('Y')
        ax.set_zlabel('Z')
        ax.set_xlim([x_range[0], x_range[1]])
        ax.set_ylim([-2, 2])
        ax.set_zlim([-2, 2])

def plot_circular_sections(func_np, x_range):
    """Plota as seções circulares no plano 2D."""
    
    x_vals = np.linspace(x_range[0], x_range[1], 10)  # Menos pontos para melhor visualização das seções
    y_vals = func_np(x_vals)

    fig, ax = plt.subplots(figsize=(8, 8))
    
    for i in range(len(x_vals)):
        circle_x = y_vals[i] * np.cos(np.linspace(0, 2 * np.pi, 100))
        circle_y = y_vals[i] * np.sin(np.linspace(0, 2 * np.pi, 100))
        
        ax.plot(circle_x + x_vals[i], circle_y + y_vals[i], color='red', alpha=0.3)

    ax.set_title('Seções Circulares no Sólido de Revolução')
    ax.set_xlabel('Eixo X')
    ax.set_ylabel('Eixo Y')

def main():
    print("Calculadora de Sólidos de Revolução")
    
    # Definir a variável simbólica
    x = sp.symbols('x')

    # Inserir manualmente a função que deseja integrar
    func_input = input("Digite a função f(x) para integrar (ex: sqrt(sin(x)) + cos(x)**2): ")
    
    try:
        func_sympy = sp.sympify(func_input)
        func_np = sp.lambdify(x, func_sympy, 'numpy')  # Converter para função NumPy
        derivative_func_sympy = sp.diff(func_sympy)   # Derivada da função
        integral_func_sympy = sp.integrate(func_sympy) # Integral da função
        
        print(f"\nFunção: f(x) = {func_sympy}")
        print(f"Derivada: f'(x) = {derivative_func_sympy}")
        print(f"Integral: ∫f(x) dx = {integral_func_sympy} + C")
        
    except sp.SympifyError as e:
        print(f"Erro na entrada da função: {e}")
        return

    # Definir o intervalo de integração
    lower_limit = float(input("Digite o limite inferior da integral: "))
    upper_limit = float(input("Digite o limite superior da integral: "))
    
    plot_area_under_curve(func_np, (lower_limit, upper_limit))
    
    plot_solid_of_revolution(func_np, (lower_limit, upper_limit))
    
   # animate_solid(func_np,(lower_limit , upper_limit))

    plot_circular_sections(func_np,(lower_limit , upper_limit))

if __name__ == "__main__":
   main()
```
