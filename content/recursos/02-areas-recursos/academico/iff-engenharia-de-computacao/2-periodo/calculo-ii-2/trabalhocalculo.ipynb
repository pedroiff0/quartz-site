## Técnicas para encontrar o volume de um sólido de revolução
1. Eixo X:
pi * integral(f(x)²,a,b)

2. Eixo Y:
Quando a reta x=L for o próprio eixo y: 2*pi*integral(x*f(x),a,b)
Quando x=L for uma assintota verical com L<=a: 2*pi*integral((x-L)*f(x),a,b)

```python
import sympy as sp
import numpy as np

# Definindo a variável simbólica
x = sp.symbols('x')

# Definindo a função
f_expr = sp.sin(x)

# Calculando o volume em torno do eixo X
volume_x = sp.integrate(sp.pi * f_expr**2, (x, 0, 2 * sp.pi)).evalf()
volume_x

# # Cálculo do volume em torno do eixo Y
# volume_y = 2 * sp.pi * sp.integrate(sp.asin(y_values), (y_values, 0, 1)).evalf()
# volume_y

# Volume em torno do eixo X
volume_x_value = volume_x.evalf()

# Volume em torno do eixo Y (considerando o intervalo correto)
# y_values_range = np.linspace(0, 1, 100)
# volume_y_value = 2 * np.pi * np.trapz(np.arcsin(y_values_range), y_values_range)

volume_x_value
```

Exemplo de Uso Eixo Y: Cascas Cilindricas:

```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
import ipywidgets as widgets
from IPython.display import display

# Configuração do Matplotlib para Jupyter
%matplotlib widget  
# Para gráficos interativos

# Definindo a variável simbólica
x = sp.symbols('x')

# Definindo a função
f_expr = 4*x - 2*x**2

# Criando um array de valores dentro do intervalo
x_values = np.linspace(0, 2, 100)
y_values = 4*x_values - 2*x_values**2

# Cálculo do volume em torno da reta x = -1 usando o método das cascas cilíndricas
volume_y = 2 * sp.pi * sp.integrate((x + 1) * f_expr, (x, 0, 2)).evalf()

# Exibindo o resultado do volume
print(f"Volume do sólido de revolução em torno da reta x=-1: {volume_y}")

# Gráfico 3D do sólido girando com controle interativo
fig = plt.figure(figsize=(10, 5))
ax = fig.add_subplot(111, projection='3d')

# Slider para controlar o ângulo de rotação
angle_slider = widgets.FloatSlider(value=0, min=0, max=360, step=1, description='Ângulo (graus):')

def plot_solid(angle):
    ax.cla()  # Limpa o gráfico anterior
    ax.set_title('Superfície de Revolução')
    ax.set_xlabel('X')
    ax.set_ylabel('Y')
    ax.set_zlabel('Z')

    # Cálculo das coordenadas do sólido girando em torno do eixo Y
    theta = np.linspace(0, angle * (np.pi / 180), 100)  # Converte graus para radianos
    X = np.outer(np.ones_like(theta), y_values)   # Rotaciona em torno do eixo Y
    Y = np.outer(np.cos(theta), y_values)          # Rotaciona em torno do eixo X baseado no ângulo
    Z = np.outer(np.sin(theta), y_values)          # Rotaciona em torno do eixo X

    # Plota a superfície até o ângulo especificado
    ax.plot_surface(X, Y, Z, color='lightgreen', alpha=0.7)

def on_angle_change(change):
    plot_solid(change['new'])

angle_slider.observe(on_angle_change, names='value')

# Exibir o slider e o gráfico juntos
display(angle_slider)

# Plot inicial com ângulo zero
plot_solid(0)
```

## Questão 1
a) y² = x³, x=4 y=0 

```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
import ipywidgets as widgets
from IPython.display import display

# Configuração do Matplotlib para Jupyter
%matplotlib widget  
# Para gráficos interativos

# Definindo a variável simbólica
x = sp.symbols('x')

# Função para a parte a
f_expr_a = sp.sqrt(x**3)

# Cálculo do volume em torno do eixo Y usando o método das cascas cilíndricas
volume_a_expr_y = sp.integrate(x * f_expr_a, (x, 0, 4))
volume_a_y = (2 * sp.pi * volume_a_expr_y).evalf()

# Cálculo do volume em torno do eixo X usando o método dos discos
volume_a_expr_x = sp.integrate(f_expr_a**2, (x, 0, 4))
volume_a_x = (sp.pi * volume_a_expr_x).evalf()

# Resultados
print(f"Volume da parte (a) em torno do eixo Y: {volume_a_y}")
print(f"Volume da parte (a) em torno do eixo X: {volume_a_x}")

# Gráfico da curva em 2D
x_vals_a = np.linspace(0, 4, 100)
y_vals_a = np.sqrt(x_vals_a**3)

plt.figure(figsize=(10,5))
plt.plot(x_vals_a, y_vals_a, label='y = √(x³)', color='blue')
plt.fill_between(x_vals_a, y_vals_a, color='lightgreen', alpha=0.5)
plt.axhline(0, color='black', lw=0.5)
plt.axvline(0, color='black', lw=0.5)
plt.title('Área sob a Curva y² = x³')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid()
plt.show()

# Gráfico do sólido girando em torno do eixo Y com controle interativo
def plot_solid_y(angle):
    fig = plt.figure(figsize=(10, 5))
    ax = fig.add_subplot(111, projection='3d')
    
    theta = np.linspace(0, angle * (np.pi / 180), num=100)
    
    X_upper = np.outer(np.ones_like(theta), np.linspace(0, 4, num=100))
    Y_upper = np.outer(np.cos(theta), [f_expr_a.subs(x,val).evalf() for val in np.linspace(0, 4, num=100)])
    Z_upper = np.outer(np.sin(theta), [f_expr_a.subs(x,val).evalf() for val in np.linspace(0, 4, num=100)])

    ax.plot_surface(X_upper,Y_upper,Z_upper,color='lightgreen', alpha=0.7)
    
    ax.set_title('Sólido de Revolução - Parte A (Eixo Y)')
    ax.set_xlabel('X')
    ax.set_ylabel('Y')
    ax.set_zlabel('Z')

# Slider para a parte a (eixo Y)
angle_slider_y_a = widgets.FloatSlider(value=0, min=0, max=360, step=1, description='Ângulo (graus):')

def on_angle_change_y_a(change):
    plot_solid_y(change['new'])

angle_slider_y_a.observe(on_angle_change_y_a,names='value')

# Exibir o slider e o gráfico inicial juntos para o eixo Y
display(angle_slider_y_a)
plot_solid_y(0)

# Gráfico do sólido girando em torno do eixo X com controle interativo
def plot_solid_x(angle):
    fig = plt.figure(figsize=(10, 5))
    ax = fig.add_subplot(111, projection='3d')
    
    theta = np.linspace(0, angle * (np.pi / 180), num=100)

    # Para a função f(x) girando em torno do eixo X
    X_upper_x = np.linspace(0, 4, num=100)
    Y_upper_x = f_expr_a.subs(x,X_upper_x).evalf()
    
    # Criar uma malha para a rotação
    X_mesh_x = np.outer(X_upper_x, np.ones_like(theta))
    Y_mesh_x = np.outer(Y_upper_x, np.cos(theta))
    Z_mesh_x = np.outer(np.ones_like(Y_upper_x), np.sin(theta))

    ax.plot_surface(X_mesh_x,Y_mesh_x,Z_mesh_x,color='lightcoral', alpha=0.7)

    ax.set_title('Sólido de Revolução - Parte A (Eixo X)')
    ax.set_xlabel('X')
    ax.set_ylabel('Y')
    ax.set_zlabel('Z')

# Slider para a parte a (eixo X)
angle_slider_x_a = widgets.FloatSlider(value=0, min=0, max=360, step=1, description='Ângulo (graus):')

def on_angle_change_x_a(change):
    plot_solid_x(change['new'])

angle_slider_x_a.observe(on_angle_change_x_a,names='value')

# Exibir o slider e o gráfico inicial juntos para o eixo X
display(angle_slider_x_a)
plot_solid_x(0)
```
