```python
x = list(range(20))
y  = [v % 5 for v in x]
plt.plot(x, y)
plt.title('Resto da divisão por 5') #titulo
plt.xlabel('Dividendo')
plt.ylabel('Resto');
```

```python
z = [v//5 for v in x]
plt.plot(x, y, label='Resto')
plt.plot(x, z, label='Quociente')
plt.title('Resto da divisão por 5')
plt.xlabel('Dividendo')
plt.legend();
```

```python
paises = ['Eua', 'Ind', 'Bra', 'Fra', 'Rus', 'Esp', 'Arg', 'Run', 'Col', 'Mex']
casos = [10.02, 8.55, 5.66, 1.84, 1.76, 1.33, 1.24, 1.2, 1.14, 0.97]

plt.bar(paises, casos, color='b', label='Confirmados')
plt.title('Casos de Covid 19 a 8/11')
plt.ylabel("Milhões")
plt.xticks(rotation='vertical')
plt.legend();
```

```python
paises = ['Eua', 'Ind', 'Bra', 'Fra', 'Rus', 'Esp', 'Arg', 'Run', 'Col', 'Mex']
casos = [10.02, 8.55, 5.66, 1.84, 1.76, 1.33, 1.24, 1.2, 1.14, 0.97]
ativos = [5.82, 0.51, 0.35, 1.66, 0.41, 1.14, 0.15, 1.14, 0.07, 0.05]

plt.bar(paises, casos, color='b', label='Confirmados')
plt.title('Casos de Covid 19 a 8/11')
plt.ylabel("Milhões")
plt.xticks(rotation='vertical')

plt.bar(paises, ativos, color='y', label='Ativos', alpha=0.9)
plt.legend();
```

```python
paises.append('outros')
casos.append(16.74)
plt.pie(casos, labels=paises, autopct = '%.1f%%', shadow=True, explode=(0, 0, 0, .6, 0, .4, 0, .2, 0, 0, .05))
plt.title("Casos Covid-19 a 8/11/2020");
```

```python
x = np.random.randint(1, 9, 10); print('Absicissas:', x)
y = np.random.randint(1, 49, 10); print('Ordenadas:', y)

plt.plot(x, y, 'or') # o = marker / r = red
plt.axis([0, 10, 0, 50]); #[xmin, xmax, ymin, ymax]
```

```python
x1 = np.arange(1, 20) / 2
y1 = x1 * 5
x2 = np.random.random(100) * 5 + 5
y2 = np.random.random(100) * 25

plt.plot(x, y, 'or', x1, y1, '*g', x2, y2, '.b')
plt.axis([0, 10, 0, 50]);
```

```python
cores = np.random.random(100)
areas = (np.random.random(100) * 30) ** 2
plt.scatter(x2, y2, c=cores, s=areas, alpha=0.5);
```
